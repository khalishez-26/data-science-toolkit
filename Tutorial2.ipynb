1. Create a numpy array of size 2 x 2

```python
#numpy is a python library that allow you to work with large datasets etc.

import numpy as np

array_2x2 = np.array ([
    [1,2],
    [3,4]
])

print("This is my numpy 2x2 array")
print(array_2x2)
```

2. Create a random numpy array of size 3 x 4

```python
#this allow you to make an array of shape 3,4 & fill it with random number. the number will be float

random3x4 = np.random.rand(3,4)

print("This is my random 3x4 array")
print(random3x4)
```

3. Create a numpy array of ones and a numpy array of zeroes

```python
ones_arrar = np.ones((3,3))
zeros_array = np.zeros((3,3))

print("Array of ones:")
print(ones_arrar)

print("Array of zeros")
print(zeros_array)
```

4. Create a random numpy array of size 6x3 and transpose it

```python
random6x3 = np.random.rand(6,3)

print("This is my randon 6x3 array:\n")
print(random6x3)

tranposed_arr = random6x3.T

print("\nThe Tranposed array:\n")
print(tranposed_arr)
```

5. Create two random numpy array of size n x p and p x m and then compute the dot product


```python
n = 2
p = 3
m = 4

random_nXp = np.random.rand(n,p)
random_pXm = np.random.rand(p,m)

dot_product = np.dot(random_nXp, random_pXm)

print("This is my random n x p array:\n")
print(random_nXp)

print("\nThis is my random p x m array:\n")
print(random_pXm)

print("\nThis is my dot product:\n")
print(dot_product)
```
